# 📊 Student Success Intelligence System

## 🎯 Objective
To analyze student performance using data analytics and predict results using machine learning.

## ⚙️ Tools Used
- Python
- Pandas
- Matplotlib
- Seaborn
- Scikit-learn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.DataFrame({
    "Name": ["Asha","Ravi","Priya","Kiran","Meena"],
    "Maths": [85,70,95,60,88],
    "Physics": [78,65,88,55,80],
    "Chemistry": [90,60,92,58,85],
    "StudyHours": [4,2,6,1,5],
    "Attendance": [90,75,95,60,85],
    "SleepHours": [7,5,8,4,6]
})

data
```

```python
data["Total"] = data[["Maths","Physics","Chemistry"]].sum(axis=1)
data["Percentage"] = data["Total"] / 3

data
```

```python
def suggestion(row):
    if row["StudyHours"] < 3:
        return "Increase study time"
    elif row["Attendance"] < 70:
        return "Improve attendance"
    elif row["SleepHours"] < 6:
        return "Maintain healthy sleep"
    else:
        return "Keep up the good work"

data["Suggestion"] = data.apply(suggestion, axis=1)

data
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np # Import numpy for select_dtypes

# Study vs Performance
plt.scatter(data["StudyHours"], data["Percentage"])
plt.xlabel("Study Hours")
plt.ylabel("Percentage")
plt.title("Study vs Performance")
plt.show()

# Correlation Heatmap
sns.heatmap(data.select_dtypes(include=np.number).corr(), annot=True)
plt.title("Correlation Analysis")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X = data[["StudyHours","Attendance","SleepHours"]]
y = data["Percentage"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LinearRegression()
model.fit(X_train, y_train)

pred = model.predict(X_test)
print("Predicted Marks:", pred)
```

## 📈 Insights
- Students who study more tend to score higher marks
- Attendance plays a key role in performance
- Proper sleep also improves results
